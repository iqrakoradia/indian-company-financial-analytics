```python
growth_test = df.sort_values(["Company", "Year"]).copy()

growth_test["Previous_Sales"] = (
    growth_test.groupby("Company")["Sales"].shift(1)
)

growth_test["Calculated_Revenue_Growth"] = np.where(
    growth_test["Previous_Sales"] != 0,
    (growth_test["Sales"] - growth_test["Previous_Sales"])
    / growth_test["Previous_Sales"],
    np.nan
)

growth_test["Revenue_Growth_Difference"] = np.where(
    growth_test["Previous_Sales"] != 0,
    growth_test["Revenue_Growth"]
    - growth_test["Calculated_Revenue_Growth"],
    np.nan
)

valid = growth_test["Revenue_Growth_Difference"].notna()

print("Revenue Growth validation")
print("=" * 60)
print("Valid comparisons:", valid.sum())
print(
    "Maximum absolute difference:",
    growth_test.loc[valid, "Revenue_Growth_Difference"].abs().max()
)
print(
    "Mean absolute difference:",
    growth_test.loc[valid, "Revenue_Growth_Difference"].abs().mean()
)
```

```python
growth_test["Previous_Net_Profit"] = (
    growth_test.groupby("Company")["Net profit"].shift(1)
)

growth_test["Calculated_Profit_Growth"] = np.where(
    growth_test["Previous_Net_Profit"] != 0,
    (
        growth_test["Net profit"]
        - growth_test["Previous_Net_Profit"]
    ) / growth_test["Previous_Net_Profit"],
    np.nan
)

growth_test["Profit_Growth_Difference"] = np.where(
    growth_test["Previous_Net_Profit"] != 0,
    growth_test["Profit_Growth"]
    - growth_test["Calculated_Profit_Growth"],
    np.nan
)

valid = growth_test["Profit_Growth_Difference"].notna()

print("Profit Growth validation")
print("=" * 60)
print("Valid comparisons:", valid.sum())
print(
    "Maximum absolute difference:",
    growth_test.loc[valid, "Profit_Growth_Difference"].abs().max()
)
print(
    "Mean absolute difference:",
    growth_test.loc[valid, "Profit_Growth_Difference"].abs().mean()
)
```

```python
growth_test["Previous_Cash"] = (
    growth_test.groupby("Company")["Cash & Bank"].shift(1)
)

growth_test["Calculated_Cash_Growth"] = np.where(
    growth_test["Previous_Cash"] != 0,
    (
        growth_test["Cash & Bank"]
        - growth_test["Previous_Cash"]
    ) / growth_test["Previous_Cash"],
    np.nan
)

growth_test["Cash_Growth_Difference"] = np.where(
    growth_test["Previous_Cash"] != 0,
    growth_test["Cash_Growth"]
    - growth_test["Calculated_Cash_Growth"],
    np.nan
)

valid = growth_test["Cash_Growth_Difference"].notna()

print("Cash Growth validation")
print("=" * 60)
print("Valid comparisons:", valid.sum())
print(
    "Maximum absolute difference:",
    growth_test.loc[valid, "Cash_Growth_Difference"].abs().max()
)
print(
    "Mean absolute difference:",
    growth_test.loc[valid, "Cash_Growth_Difference"].abs().mean()
)
```

```python
cagr_test = df.sort_values(["Company", "Year"]).copy()

cagr_test["Revenue_3Y_Ago"] = (
    cagr_test.groupby("Company")["Sales"].shift(3)
)

cagr_test["Calculated_Revenue_CAGR_3Y"] = np.where(
    (cagr_test["Revenue_3Y_Ago"] > 0) & (cagr_test["Sales"] > 0),
    (
        (cagr_test["Sales"] / cagr_test["Revenue_3Y_Ago"]) ** (1 / 3)
    ) - 1,
    np.nan
)

cagr_test["Revenue_CAGR_3Y_Difference"] = np.where(
    cagr_test["Calculated_Revenue_CAGR_3Y"].notna(),
    cagr_test["Revenue_CAGR_3Y"]
    - cagr_test["Calculated_Revenue_CAGR_3Y"],
    np.nan
)

valid = cagr_test["Revenue_CAGR_3Y_Difference"].notna()

print("Revenue CAGR 3Y validation")
print("=" * 60)
print("Valid comparisons:", valid.sum())
print(
    "Maximum absolute difference:",
    cagr_test.loc[valid, "Revenue_CAGR_3Y_Difference"].abs().max()
)
print(
    "Mean absolute difference:",
    cagr_test.loc[valid, "Revenue_CAGR_3Y_Difference"].abs().mean()
)
```
